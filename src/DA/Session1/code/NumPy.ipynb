```python
def array_reshape(arr : list ,rows : int, cols : int):
    try:
        arr = np.array(arr)
        return arr.reshape(rows, cols)
    except(TypeError,ValueError) as e:
        raise ValueError(f"Invalid reshape:{e}")
```

```python
vec = [1,2,3,4,5,6]
array_reshape(vec,2,3)
```
