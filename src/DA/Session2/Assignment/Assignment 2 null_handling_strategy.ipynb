# Assignment 2: The `null_handling_strategy` Function

### Scenario:
Incoming user data often has missing values. We must implement a flexible strategy to handle these "Null Values" to prepare data for Machine Learning pipelines.

### Requirements:
1. Check for null values in the DataFrame.
2. Apply a strategy based on parameters: `"drop_rows"`, `"fill_mean"`, or `"fill_median"`.
3. Ensure the function only fills numerical columns when using mean or median.

```python
import pandas as pd
import numpy as np

# 1. Dataset Creation for Assignment 2
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home', 
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160],
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40], # Contains Nulls (NaN)
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)
df_test.head()
```

```python
def null_handling_strategy(df, strategy="fill_mean"):
    """
    Company Task: Clean a dataset by resolving missing (NaN) values.
    
    Parameters:
    - df: Input DataFrame
    - strategy: "drop_rows", "fill_mean", or "fill_median"
    
    Returns:
    - Cleaned DataFrame copy
    """
    # Create an explicit copy to avoid modifying the original dataframe
    df_cleaned = df.copy()
    
    # 1. Strategy: Drop rows with missing values
    if strategy == "drop_rows":
        df_cleaned = df_cleaned.dropna()
        
    # 2. Strategy: Fill missing values with column Mean (Numerical only)
    elif strategy == "fill_mean":
        num_cols = df_cleaned.select_dtypes(include="number").columns
        for col in num_cols:
            if df_cleaned[col].isnull().any():
                df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())
                
    # 3. Strategy: Fill missing values with column Median (Numerical only)
    elif strategy == "fill_median":
        num_cols = df_cleaned.select_dtypes(include="number").columns
        for col in num_cols:
            if df_cleaned[col].isnull().any():
                df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())
                
    else:
        raise ValueError(f"Unknown strategy: '{strategy}'. Supported strategies: 'drop_rows', 'fill_mean', 'fill_median'.")
        
    return df_cleaned

# --- Test Cases ---
print("Initial Nulls in Customer_Age:", df_test['Customer_Age'].isnull().sum())

# Test 1: Fill with Median
df_median = null_handling_strategy(df_test, strategy="fill_median")
print("\n--- Test 'fill_median' ---")
print("Nulls after imputation:", df_median['Customer_Age'].isnull().sum())
print(df_median[['Transaction_ID', 'Customer_Age']])

# Test 2: Drop Rows
df_dropped = null_handling_strategy(df_test, strategy="drop_rows")
print("\n--- Test 'drop_rows' ---")
print(f"Original row count: {len(df_test)} -> Cleaned row count: {len(df_dropped)}")
```
