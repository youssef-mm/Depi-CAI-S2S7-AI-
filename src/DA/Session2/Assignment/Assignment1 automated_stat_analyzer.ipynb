# Assignment 1: The `automated_stat_analyzer` Function

### Scenario:
A retail company needs a utility to quickly summarize sales data. We must create a function that identifies the **Central Tendency** and **Dispersion** of any numerical column.

### Requirements:
1. Accept a Pandas DataFrame and a column name.
2. Calculate the **Mean**, **Median**, and **Standard Deviation**.
3. Identify if the data is **Skewed** by comparing the Mean and Median.
4. **Bonus:** If the column is categorical, return the **Mode** instead.

```python
import pandas as pd
import numpy as np

# 1. Dataset Creation for Assignment 1
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home', 
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160], # 1000 is an Outlier
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],    # Contains Nulls
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)
df_test.to_csv('company_sales_test.csv', index=False)
df_test.head()
```

```python
def automated_stat_analyzer(df, column_name):
    """
    Company Task: Provide a summary report of a specific data variable.
    
    Instructions:
    1. Check if the column is numerical or categorical.
    2. For numerical: Calculate Mean, Median, and Standard Deviation.
    3. For categorical: Calculate the Mode.
    4. Return a dictionary with these statistical measures.
    """
    col_data = df[column_name]
    
    # 1. Check if the column is numerical
    if pd.api.types.is_numeric_dtype(col_data):
        mean_val = col_data.mean()
        median_val = col_data.median()
        std_val = col_data.std()
        
        # Check Skewness (Mean vs Median)
        if round(mean_val, 2) > round(median_val, 2):
            skewness = "Positively Skewed (Right-Skewed)"
        elif round(mean_val, 2) < round(median_val, 2):
            skewness = "Negatively Skewed (Left-Skewed)"
        else:
            skewness = "Symmetric"
            
        return {
            "Column": column_name,
            "Type": "Numerical",
            "Mean": mean_val,
            "Median": median_val,
            "Standard Deviation": std_val,
            "Skewness": skewness
        }
        
    # 2. Categorical Column (Bonus)
    else:
        modes = col_data.mode()
        mode_val = modes[0] if not modes.empty else None
        
        return {
            "Column": column_name,
            "Type": "Categorical",
            "Mode": mode_val
        }

# --- Test Cases ---
print("--- Testing Numerical Feature (Sales_Amount) ---")
print(automated_stat_analyzer(df_test, 'Sales_Amount'))

print("\n--- Testing Categorical Feature (Product_Category) ---")
print(automated_stat_analyzer(df_test, 'Product_Category'))
```
