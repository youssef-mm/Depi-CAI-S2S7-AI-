Cell 1: Even/Odd Checker

```python
class EvenOddChecker:
    """
    EvenOddChecker class handles the operation of checking whether a number is even or odd.

    Attributes:
        number (int): The number to be checked.
    """
    def __init__(self, number: int):
        """Initializes the checker with a target number."""
        self.number = number

    def check_even_odd(self) -> str:
        """
        Determines whether the number is even or odd.

        Returns:
            str: 'Even' if the number is even, otherwise 'Odd'.
        """
        if self.number % 2 == 0:
            return "Even"
        return "Odd"

# Testing Object
checker = EvenOddChecker(15)
print(f"The number {checker.number} is {checker.check_even_odd()}.")
```

Cell 2: Basic Calculator

```python
class BasicCalculator:
    """
    BasicCalculator class performs basic arithmetic operations.

    Attributes:
        num1 (float): The first operand.
        num2 (float): The second operand.
    """
    def __init__(self, num1: float, num2: float):
        """Initializes the calculator with two numbers."""
        self.num1 = num1
        self.num2 = num2

    def add(self) -> float:
        """Returns the sum of num1 and num2."""
        return self.num1 + self.num2

    def subtract(self) -> float:
        """Returns the difference of num1 and num2."""
        return self.num1 - self.num2

    def multiply(self) -> float:
        """Returns the product of num1 and num2."""
        return self.num1 * self.num2

    def divide(self):
        """
        Returns the division result of num1 by num2.
        Handles division by zero.
        """
        if self.num2 == 0:
            return "Error: Division by zero is not allowed."
        return self.num1 / self.num2

# Testing Object
calc = BasicCalculator(20.0, 4.0)
print(f"Addition: {calc.add()}")
print(f"Subtraction: {calc.subtract()}")
print(f"Multiplication: {calc.multiply()}")
print(f"Division: {calc.divide()}")
```

Cell 3: Multiplication Table Generator

```python
class MultiplicationTable:
    """
    MultiplicationTable class generates the multiplication table for a number.

    Attributes:
        number (int): The base number.
        limit (int): The upper multiplier limit.
    """
    def __init__(self, number: int, limit: int = 10):
        """Initializes the base number and multiplier limit."""
        self.number = number
        self.limit = limit

    def display_table(self) -> None:
        """Prints the multiplication table from 1 to the limit."""
        for i in range(1, self.limit + 1):
            print(f"{self.number} * {i} = {self.number * i}")

# Testing Object
table = MultiplicationTable(9)
table.display_table()
```

Cell 4: Twin Primes Finder

```python
class TwinPrimesFinder:
    """
    TwinPrimesFinder class identifies twin primes within a given boundary.

    Attributes:
        limit (int): Upper bound limit for prime search.
    """
    def __init__(self, limit: int = 1000):
        """Initializes the search limit."""
        self.limit = limit

    def is_prime(self, n: int) -> bool:
        """Helper method to check if an integer is prime."""
        if n <= 1:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True

    def find_twin_primes(self) -> list:
        """
        Finds all twin prime pairs below the upper limit.

        Returns:
            list: List of tuples containing twin prime pairs.
        """
        twins = []
        for i in range(3, self.limit - 2, 2):
            if self.is_prime(i) and self.is_prime(i + 2):
                twins.append((i, i + 2))
        return twins

# Testing Object
finder = TwinPrimesFinder(100)
for p1, p2 in finder.find_twin_primes():
    print(f"{p1} and {p2}")
```

Cell 5: Prime Factorizer

```python
class PrimeFactorizer:
    """
    PrimeFactorizer class calculates prime factors of an integer.

    Attributes:
        number (int): The target integer to factorize.
    """
    def __init__(self, number: int):
        """Initializes the factorizer with a target number."""
        self.number = number

    def get_prime_factors(self) -> list:
        """
        Calculates prime factors of the initialized number.

        Returns:
            list: Prime factors list.
        """
        n = self.number
        factors = []
        while n % 2 == 0:
            factors.append(2)
            n //= 2
            
        for i in range(3, int(n ** 0.5) + 1, 2):
            while n % i == 0:
                factors.append(i)
                n //= i
                
        if n > 2:
            factors.append(n)
        return factors

# Testing Object
factorizer = PrimeFactorizer(56)
print(f"Prime factors of 56: {factorizer.get_prime_factors()}")
```

Cell 6: Decimal to Binary Converter

```python
class DecimalToBinary:
    """
    DecimalToBinary class handles conversion from base 10 to base 2.

    Attributes:
        number (int): The decimal number to convert.
    """
    def __init__(self, number: int):
        """Initializes the decimal number."""
        self.number = number

    def convert(self) -> str:
        """
        Converts the decimal value to a binary string representation.

        Returns:
            str: Binary string.
        """
        if self.number == 0:
            return "0"
        
        n = self.number
        binary = ""
        while n > 0:
            binary = str(n % 2) + binary
            n //= 2
        return binary

# Testing Object
converter = DecimalToBinary(11)
print(f"Binary of {converter.number}: {converter.convert()}")
```

Cell 7: Perfect Numbers Finder

```python
class PerfectNumbersFinder:
    """
    PerfectNumbersFinder class finds perfect numbers in a specific range.

    Attributes:
        start (int): Start of the search range.
        end (int): End of the search range.
    """
    def __init__(self, start: int = 0, end: int = 100):
        """Initializes the start and end boundary range."""
        self.start = start
        self.end = end

    def is_perfect(self, n: int) -> bool:
        """Helper method to determine if a number equals sum of its proper divisors."""
        if n <= 1:
            return False
        return sum(i for i in range(1, n) if n % i == 0) == n

    def get_perfect_numbers(self) -> list:
        """
        Collects all perfect numbers in the given range.

        Returns:
            list: List of perfect numbers.
        """
        perfect_nums = []
        for num in range(self.start, self.end + 1):
            if num == 0:
                perfect_nums.append(0)
            elif self.is_perfect(num):
                perfect_nums.append(num)
        return perfect_nums

# Testing Object
p_finder = PerfectNumbersFinder(0, 100)
for num in p_finder.get_perfect_numbers():
    print(num)
```

Cell 8: Word Frequency Counter

```python
class WordFrequencyCounter:
    """
    WordFrequencyCounter class tracks word occurrences in a dataset.

    Attributes:
        words (list): List of strings.
    """
    def __init__(self, words: list):
        """Initializes the counter with a word list."""
        self.words = words

    def get_frequencies(self) -> dict:
        """
        Counts the occurrences of each word.

        Returns:
            dict: Words mapped to their frequency counts.
        """
        counts = {}
        for word in self.words:
            counts[word] = counts.get(word, 0) + 1
        return counts

# Testing Object
sample_list = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
word_counter = WordFrequencyCounter(sample_list)
print("Frequency Dictionary:", word_counter.get_frequencies())
```

Cell 9: Rule-Based Chatbot

```python
import random

class RuleBasedChatbot:
    """
    RuleBasedChatbot class provides pre-configured replies based on detected keywords.

    Attributes:
        responses (dict): Intent and keyword mappings for responses.
    """
    def __init__(self, responses: dict = None):
        """Initializes default conversational patterns or accepts custom mappings."""
        if responses is None:
            self.responses = {
                "hello": ["Hello!", "Hi there!", "Greetings!"],
                "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
                "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
                "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
            }
        else:
            self.responses = responses

    def get_reply(self, message: str) -> str:
        """
        Generates a suitable response for a given text prompt.

        Returns:
            str: Selected response string.
        """
        cleaned_msg = message.lower()
        for key in self.responses:
            if key != "default" and key in cleaned_msg:
                return random.choice(self.responses[key])
        return random.choice(self.responses["default"])

# Testing Object
chatbot = RuleBasedChatbot()
print("Bot response:", chatbot.get_reply("Hello, how are you?"))
```
