```python
def add(x,y):
    return x + y

class calc:
    def __init__(self,x,y) -> None:
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y     

c1 = calc(10,20)       
print(c1.add())
```

```python
from abc import ABC,abstractmethod

class animal(ABC):
    @abstractmethod
    def move(self):
        pass

class bird(animal):

    def move(self):
        print("move from bird")    

class cat(animal):

    def move(self):
        print("move from cat")

a = bird()
a.move()       

b = cat()
b.move()
```

```python
class rectangle:
    __length = 6
    __width = 12

    def area(self):
        return self.__length * self.__width

r = rectangle()
print(r.area())
print(r._rectangle__length)
print(r._rectangle__width)
```

```python
class rectangle:

    def __init__(self,length,width):
        self.__length = length
        self.width = width

    @property
    def length(self):
        return self.__length

    @length.setter
    def length(self, value):
        if value <=0:
            raise ValueError("Length must be positive")
        self.__length = value

r = rectangle(6,12)
print(r.length)        
```

```python
class shape:
    def calc_area(self):
        pass

class circle(shape):
    def __init__(self,radius) -> None:
        self.radius = radius

    def calc_area(self):
        return 3.14 * self.radius * self.radius

class sqare(shape):
    def __init__(self,side) -> None:
        self.side = side

    def calc_area(self):
        return self.side**2

x = circle(5)
y = sqare(4)
print("Area : ",x.calc_area())    
print("Area : ",y.calc_area())    
```

```python
class dog:
    def speak(self):
        return "woof"

class cat:
    def speak(self):
        return "meow"

d1 = dog()
c1 = cat()

def make_sound(animal):
    return animal.speak()

print(make_sound(d1))
print(make_sound(c1))
```

```python
class model:

    def fit(self,x,y):
        raise NotImplementedError

    def predict(self,x):
        raise NotImplementedError

class linear_regression(model):
    def fit(self,x,y):
        print("fit from linear regression")

    def predict(self,x):
        print("predict from linear regression")

class decision_tree(model):
    def fit(self,x,y):
        print("fit from decision tree")

    def predict(self,x):
        print("predict from decision tree")

models = [linear_regression(),decision_tree()]
for model in models:
    model.fit(1,2)
    model.predict(1)                
```
