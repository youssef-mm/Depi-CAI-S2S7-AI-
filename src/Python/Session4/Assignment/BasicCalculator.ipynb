```python
import operator
ops = {
    '1': ('+', operator.add),
    '2': ('-', operator.sub),
    '3': ('*', operator.mul),
    '4': ('/', operator.truediv)
}
def calculator():
    print("1. Add (+)\n2. Subtract (-)\n3. Multiply (*)\n4. Divide (/)")

    while True:
        choice = input("\nEnter choice (1-4) or 'exit': ").strip().lower()
        if choice == 'exit':
            print("Goodbye!")
            break

        if choice in ops:
            try:
                n1 = float(input("First number: "))
                n2 = float(input("Second number: "))
                
                if choice == '4' and n2 == 0:
                    print("Error: Cannot divide by zero!")
                    continue

                symbol, func = ops[choice]
                print(f"{n1} {symbol} {n2} = {func(n1, n2)}")
            except ValueError:
                print("Invalid number!")
        else:
            print("Invalid choice!")

calculator()
```
