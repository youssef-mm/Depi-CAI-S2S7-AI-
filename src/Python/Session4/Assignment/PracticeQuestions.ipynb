1) Write a function that inputs a number and prints the multiplication table of that number 

```python
def print_multiplication_table():
    num = int(input("Enter a number to print its multiplication table: "))

    for i in range(1,11):
        result = num * i
        print(f"{num} x {i} = {result}")

print_multiplication_table()
```

2) Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes 

```python
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True

def print_twin_primes(limit):
    print(f"Twin primes up to {limit}:")
    for num in range(2, limit - 1):
        if is_prime(num) and is_prime(num + 2):
            print(f"({num}, {num + 2})")

print_twin_primes(1000)            
    
```

3) Write a program to find out the prime factors of a number. 

```python
def find_prime_factors(n):
    factors = []
    for i in range(2, n + 1):
        while n % i == 0:
            factors.append(i)
            n //= i
    return factors

find_prime_factors(56)
```

4) Write a function that converts a decimal number to binary number 

```python
def decimal_to_binary(n):
    if n == 0:
        return "0"
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary
        n //= 2
    return binary

decimal_to_binary(11)
```

 5) A number is called perfect if the sum of proper divisors of that number is equal to the number. For example 28 is perfect number, since 1+2+4+7+14=28. Write a program to print all the perfect numbers in a given range 

```python
def perfectNums(start, end):
    perfect_numbers = []
    for num in range(start, end + 1):
        divisors_sum = sum(i for i in range(1, num) if num % i == 0)
        if divisors_sum == num:
            perfect_numbers.append(num)
    return perfect_numbers

perfectNums(0, 100)    
```
