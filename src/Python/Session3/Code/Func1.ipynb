```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"Training model: {model_name}")
    print("Datasets:")
    for dataset in datasets:
        print(f"- {dataset}")
    print("Hyperparameters:")
    for key, value in hyperparameters.items():
        print(f"- {key}: {value}")
```

```python
train_model("fraud_detection", "dataset1.csv", "dataset2.csv", learning_rate=0.01, epochs=10, batch_size=32, optimizer="Youssef")
```
