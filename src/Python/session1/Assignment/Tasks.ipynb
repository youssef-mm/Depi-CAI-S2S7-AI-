## Task 1

```python
email = input("Enter your email address: ").strip()
if email.count("@") == 1 and "." in email.split("@")[1]:
    extract_username = email.split("@")[0]
    domain_part = email.split("@")[1]
    extracted_domain = domain_part[:domain_part.rfind(".")]
    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"  
    print(f"Username: {extract_username}")
    print(f"Domain: {extracted_domain}")
    print(f"Domain Type: {domain_type}")
else:
    print("Invalid email")
```
